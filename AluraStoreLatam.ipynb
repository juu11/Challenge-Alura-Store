Resumen de ventas por categorias

```python
tiendas = {"Tienda 1": tienda,
            "Tienda 2": tienda2,
            "Tienda 3": tienda3,
           "Tienda 4": tienda4
          }

ventas_categoria= pd.concat(
                              {
                                  tienda: df.groupby("Categoría del Producto").size()
                                  for tienda, df in tiendas.items()
                              },axis=1
                            ).fillna(0).astype(int)

ventas_categoria["Total"] = ventas_categoria.sum(axis=1)
ventas_categoria = ventas_categoria.sort_values(
                                                by="Total",
                                              ascending=False
                                                )

ventas_categoria = ventas_categoria.reset_index()
ventas_categoria.rename(columns={"Categoría del Producto": "Categoría de Productos"},
                        inplace=True)
display(ventas_categoria)
```

# 4. Productos más y menos vendidos

Tienda 1

producto más vendido

```python
mas_vendidos = []
menos_vendidos = []

for i, tiendad in enumerate([tienda,tienda2,tienda3,tienda4]):
  nombre_tienda = f'Tienda {i+1}'
  producto = tiendad['Producto'].value_counts().reset_index()
  producto.columns = ['Producto', 'Cantidad']

  #mas vendidos
  M_producto = producto.head(5).copy()
  M_producto.loc[:, 'Tienda'] = nombre_tienda
  M_producto.loc[:, 'Tipo'] = 'Más Vendidos'
  mas_vendidos.append(M_producto)

  #menos vendidos
  m_producto = producto.tail(5).copy()
  m_producto.loc[:, 'Tienda'] = nombre_tienda
  m_producto.loc[:, 'Tipo'] = 'Menos Vendidos'
  menos_vendidos.append(m_producto)

A_mas_vendidos = pd.concat(mas_vendidos,ignore_index=True)
A_menos_vendidos = pd.concat(menos_vendidos,ignore_index=True)

print('Productos más vendidos por tiendas\n')
for i in mas_vendidos:
  nombre_tienda = i['Tienda'].iloc[0]
  print(f'{nombre_tienda}')
  display(i[['Producto', 'Cantidad']])
  print('\n')

print('Productos menos vendidos por tiendas\n')
for i in menos_vendidos:
  nombre_tienda = i['Tienda'].iloc[0]
  print(f'{nombre_tienda}')
  display(i[['Producto', 'Cantidad']])
  print('\n')

```

```python
data_envios = pd.DataFrame({
    'Tienda': ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4'],
    'Envios': [promedio_1, promedio_2, promedio_3, promedio_4]
})
plt.figure(figsize=(10, 6))
plt.pie(
    data_envios['Envios'],
    labels=data_envios['Tienda'],
    autopct='%1.1f%%',
    startangle=90,
    colors=['red', '#D8E6C8', '#6CAF49', '#3E7D31'],
    explode=(0.05, 0, 0, 0)
)
plt.title('Envio promedio por tiendda', fontsize=14, weight='bold')
plt.show()
```
